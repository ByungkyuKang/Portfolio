# 07. Divorce Rates by Religion: Do religious people divorce less?

### Inspect and preprocess the religion variable

````python
# Inspection
welfare['religion'].dtype
````

````python
welfare['religion'].value_counts() # No missing value handling is required
````

````python
###############################
# Preprocessing
#
# value | Meaning 
# -----------------------------
#   1   |  yes
#   2   |  no
#   9   | Don'tknow or N/A
###############################
welfare['religion'] = np.where( welfare['religion']==1, 'yes', 'no' )
welfare['religion'].value_counts()
````

````python
# graph
sns.countplot( data=welfare, x='religion' )
````

### Inspect and preprocess the the Marital Status Variable

````python
welfare['marriage_type'].dtype
````

````python
welfare['marriage_type'].value_counts()
````

````python
# Preprocessing:
#     Creating a Derived Column: Using only 'Married/spouse present (1)' and 'Divorced (3)'

welfare['marriage'] = np.where( welfare['marriage_type']==1, 'married',
                                     np.where( welfare['marriage_type']==3, 'divorced', 'etc'))

cnt_marriage_satus = welfare.groupby( 'marriage', as_index=False ).\
    agg( n_marriage_status=('marriage','count') )

cnt_marriage_satus
````

````python
# graph
sns.barplot( data=cnt_marriage_satus, x='marriage', y='n_marriage_status', \
             order=['married', 'divorced', 'etc'] )
````

- According to the table and graph above, there are 7,190 `married` individuals, 689 `divorced` individuals, and 6,539 classified as `etc`. Since those in the `etc` category are not the subject of this analysis, they should be removed later.

## Analyze divorce rates by religious status

````python
# Create divorce rate table by religious status
religion_divorced = welfare.query('marriage != "etc"').groupby('religion', as_index=False )['marriage']\
    .value_counts( normalize=True )
religion_divorced
````

````python
religion_divorced = religion_divorced.query('marriage=="divorced"')\
    .assign( proportion=religion_divorced['proportion']*100 ).round(1)

religion_divorced
````

````python
# Creating a graph
sns.barplot( data=religion_divorced, x='religion', y='proportion' )
````

#### <b><u>Key Takeaway</u></b>: The table and graph show a divorce rate of 8.0% for religious individuals and 9.5% for non-religious individuals. While this indicates a slightly lower divorce rate among those with a religion, the 1.5 percentage point difference is relatively small, suggesting that further statistical testing is needed to determine if this variation is statistically significant.

## Analyze divorce rates by age group and religious status

````python
# Create divorce rate table by age group
age_group_divorced = welfare.query( 'marriage!="etc"' ).groupby( 'age_group', as_index=False)['marriage']\
    .value_counts( normalize=True )

age_group_divorced
````

````python
welfare.query( 'marriage!="etc"' ).groupby( 'age_group', as_index=False )['marriage'].value_counts()
````

- In the young group, everyone is married. However, since there are only two cases, it is inappropriate to include them in the analysis. Therefore, the young group is excluded from this analysis

````python
# Creating a graph
age_group_divorced = age_group_divorced.query( 'age_group!="young" & marriage=="divorced"' )\
    .assign( proportion=age_group_divorced['proportion']*100 ).round(1)

print(age_group_divorced)

sns.barplot( data=age_group_divorced, x='age_group', y='proportion' )
````

````python
# Create divorce rate table by age group and religious status
age_group_religion_divorced = welfare.query( 'marriage!="etc" & age_group!="young"' )\
     .groupby( ['age_group', 'religion'], as_index=False )['marriage'].value_counts( normalize=True )

age_group_religion_divorced
````

````python
# Plot divorce rates by age group and religious status
age_group_religion_divorced = age_group_religion_divorced.query( 'marriage=="divorced"' )\
    .assign( proportion=age_group_religion_divorced['proportion']*100 ).round(1)
````

````python
print(age_group_religion_divorced)
sns.barplot( data=age_group_religion_divorced, x='age_group', y='proportion', hue='religion' )
````

#### <b><u>Key Takeaway</u></b>: According to the table and graph above, the divorce rate for non-religious individuals is higher in both the middle and senior groups. In both age cohorts, the divorce rate for those without a religion is about 1.3 percentage points higher

### Inspection and Preprocessing for Geographic Regions

````python
# Inspection
welfare['code_region'].dtype
````

````python
welfare['code_region'].value_counts().sort_index()
````

````python
# preprocessing
list_region = pd.DataFrame(
    {
        'code_region': [1, 2, 3, 4, 5, 6, 7],
        'region': ['Seoul',
                   'Seoul metropolitan area (Incheon/Gyeonggi)',
                   'Busan/Gyeongnam/Ulsan',
                   'Daegu/Gyeongbuk',
                   'Daejeon/Chungnam',
                   'Gangwon/Chungbuk',
                   'Gwangju/Jeonnam/Jeonbuk/Jeju']
    }
)

list_region
````

````python
# adding `region` to the welfare dataframe
welfare = welfare.merge( list_region, how='left', on='code_region' )
welfare[ ['code_region', 'region'] ]
````

## Analyzing the Proportion of Age Groups by Region

````python
# Creating an Age Group Proportion Table by Region
region_age_group = welfare.groupby( 'region', as_index=False )['age_group'].value_counts( normalize=True )
````

````python
region_age_group
````

````python
# Convert to Percentage
region_age_group = region_age_group.assign( proportion=region_age_group['proportion']*100 ).round(1)
````

````python
region_age_group
````

````python
# Creating a graph
sns.barplot( data=region_age_group, x='proportion', y='region', hue='age_group' )
````

##### Plot stacked bar chart

````python
# Pivot table creation
pivot_region_age_group = region_age_group[ ['region', 'age_group', 'proportion'] ]\
    .pivot( index='region', columns='age_group', values='proportion' )

pivot_region_age_group
````

##### Creating graphs

````python
# Plot horizontal bar chart
pivot_region_age_group.plot.barh( stacked=True )
````

````python
# sorting bars by 'senior'
reorder_pivot = pivot_region_age_group.sort_values('senior')[['young', 'middle', 'senior']]
reorder_pivot
````

````python
reorder_pivot.plot.barh( stacked=True )
````

#### <b><u>Key takeaway</u></b>: According to the graph above, 'Daegu/Gyeongbuk' has the highest proportion of the elderly, followed by 'Gangwon/Chungbuk', 'Gwangju/Jeonnam/Jeonbuk/Jeju', and 'Busan/Gyeongnam/Ulsan'.

# Overall Summary and Key Findings

This analysis examined several sociodemographic characteristics in the 2019 Korean Welfare Panel Study, including gender, age, occupation, religion, marital status, and region.

## Key Findings

1. **Gender wage gap**
   - Men had a substantially higher average monthly income than women in the analyzed sample.
   - The difference was relatively small among younger adults but became much wider in the middle-aged group.

2. **Age and income**
   - Average income generally increased from early adulthood, reached its highest level during middle age, and then declined among older adults.
   - The middle-aged group had the highest average income, while the senior group had the lowest.

3. **Income differences by occupation**
   - Average income varied considerably across occupations.
   - Medical practitioners were among the highest-paid occupational groups, while several care, service, and elementary occupations were among the lowest-paid groups.
   - This indicates that occupation is strongly associated with income differences.

4. **Occupational distribution by gender**
   - Crop growers were the most common occupation for both men and women.
   - Beyond the top occupation, the occupational rankings differed by gender, suggesting that men and women were concentrated in different types of work.

5. **Religion and divorce**
   - Non-religious respondents showed a slightly higher proportion of divorced individuals than religious respondents.
   - The same pattern appeared within both the middle-aged and senior groups.
   - However, this descriptive result alone does not establish that religion directly affects divorce.

6. **Regional age-group distribution**
   - Daegu/Gyeongbuk had the highest proportion of senior residents among the seven regional groups.
   - Regional differences in age composition may reflect differences in migration, employment opportunities, and population aging.

## Overall Interpretation

The results show that income and demographic outcomes are related to multiple factors rather than a single characteristic. Gender, age, occupation, religion, marital status, and region interact with broader social and economic conditions. Therefore, the findings should be interpreted as descriptive patterns within the survey data rather than evidence of direct causal relationships.

# Important Considerations

- **Average values do not describe the entire distribution.** A small number of high-income observations may raise the mean, so medians and income distributions could be examined in a future analysis.
- **Group sizes may differ.** Results for occupations or demographic groups with few observations can be unstable and should be interpreted cautiously.
- **Missing income values were excluded.** The income analysis therefore represents respondents with available income data rather than every respondent in the survey.
- **The analysis is observational.** Differences between groups show associations, not proof that one characteristic causes another.
- **Several variables were simplified into broader categories.** For example, age and marital status were grouped to make comparison easier, but this may hide differences within each category.
- **The data represent a specific survey period.** The patterns found in the 2019 survey may not describe current conditions.

# Conclusion

This project demonstrates a complete exploratory data analysis workflow: inspecting variables, preprocessing coded survey data, creating derived variables, summarizing groups, and communicating results through visualizations. It also highlights the importance of interpreting statistical differences carefully and considering sample size, missing data, variable definitions, and the limitations of observational data.
