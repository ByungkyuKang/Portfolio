## 01. Program to sum numbers in a list

```python
# Create a list that contains numbers
num_to_sum = [10, 20, 30, 40, 50]

result = 0
for num in num_to_sum:
    result = result+num

print(result)

# Easier way for a list
print(sum(num_to_sum))
```

## 02. Even/Odd Chacker

```python
## 02. Even/Odd Checker Program
numbers = [2,6,8,3,7,11]

for number in numbers:
    if number%2==0:
        print(f"{number} is even.")
    else:
        print(f"{number} is odd.")
```
