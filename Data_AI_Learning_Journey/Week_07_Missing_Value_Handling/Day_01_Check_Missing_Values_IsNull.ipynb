```python
# Creating a data file
import os
import pandas as pd

input_dir = "../assets/Week07/Day01/Input"
output_dir = "../assets/Week07/Day01/output"

os.makedirs(input_dir, exist_ok=True)
os.makedirs(output_dir, exist_ok=True)

input_file_path = f"{input_dir}/employee_data_with_missing_values.csv"

employee_data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Henry"],
    "Department": ["Engineering", "Data", None, "Cloud", "Data", "Security", None, "Engineering"],
    "Experience": [5, 3, 7, None, 2, 8, 6, 1],
    "Score": [88, 92, None, 85, 67, 95, 90, None],
    "Status": ["Active", "Active", "Inactive", "Active", None, "Active", "Active", "Inactive"],
    "Remote": [True, True, False, True, None, False, True, True]
}

# Create a DataFrame
df_employees = pd.DataFrame(employee_data)

# Save the DataFrame as a CSV file
df_employees.to_csv(input_file_path, index=False)

print("employee_data_with_missing_values.csv has been created.")
```

```python
# Read the employee CSV file
df_employees = pd.read_csv(input_file_path)
    
# Display the DataFrame
df_employees
```

## 01. Checking for Missing Values using isnull()

```python
missing_value_table = df_employees.isnull()
missing_value_table
```

## 02. Counting missing values per column

```python
# Count missing values in each column
missing_count_by_column = df_employees.isnull().sum()

missing_count_by_column
```

## 03. Total number of missing values across the entire DataFrame

```python
# Count the total number of missing values
total_missing_count = df_employees.isnull().sum().sum()

print(f"Total missing values: {total_missing_count}")
```

## 04. Calculating the percentage of missing values per column

```python
# Calculate missing value percentages by column
missing_percentage_by_column = ( df_employees.isnull().mean() * 100 ).round(2)

missing_percentage_by_column
```

## 05. Filtering rows with at least one missing value

```python
# Find rows that contain at least one missing value
rows_with_missing_values = df_employees[ df_employees.isnull().any(axis=1) ]

rows_with_missing_values
```

## 06. Counting missing values per row

```python
# Count missing values in each row
missing_count_by_row = df_employees.isnull().sum(axis=1)

missing_count_by_row
```

```python
# Create a temporary DataFrame with missing value counts
df_missing_count = df_employees.copy()

df_missing_count["Missing_Value_Count"] = df_missing_count.isnull().sum(axis=1)

df_missing_count
```

## 07. Finding missing values in a specific column

```python
# e.g.) Find employees with missing Score values
missing_score_rows = df_employees[ df_employees["Score"].isnull() ]

missing_score_rows
```

```python
# e.g.) Find employees with missing Department values
missing_department_rows = df_employees[ df_employees["Department"].isnull() ]

missing_department_rows
```

## 08. Extracting rows with no missing values
- .notnull().all(axis=1)

```python
complete_rows = df_employees[ df_employees.notnull().all(axis=1) ]

complete_rows
```

## 09. Checking for the existence of any missing values

```python
# Check whether the DataFrame contains any missing values
has_missing_values = df_employees.isnull().any().any()

print(f"Does the DataFrame contain missing values? {has_missing_values}")
```

## 10. Creating Missing Value Summary

```python
# Create missing value counts
missing_count_by_column = df_employees.isnull().sum()

# Create missing value percentages
missing_percentage_by_column = (df_employees.isnull().mean() * 100).round(2)

# Create a missing value summary DataFrame
missing_value_summary = pd.DataFrame(
    {
        "Column":df_employees.columns,
        "Missing_Count":missing_count_by_column.values,
        "Missing_Percentage":missing_percentage_by_column.values
    }
)
missing_value_summary
```

```python
missing_value_summary_with_missing_only = missing_value_summary[ missing_value_summary["Missing_Count"] > 0 ]

missing_value_summary_with_missing_only
```

## 11. Saving the summary

```python
rows_with_missing_values
```

```python
missing_value_summary.to_csv(f"{output_dir}/missing_value_summary.csv", index=False)
rows_with_missing_values.to_csv(f"{output_dir}/rows_with_missing_values.csv", index=False)
missing_value_summary_with_missing_only.to_csv(f"{output_dir}/missing_values_only.csv", index=False)

print("Missing value output files have been saved.")
```
