## 01. Return
- A keyword that returns the output processed within a function to the caller.

```python
def add_numbers(a, b):
    # Return the sum of two numbers
    return a + b


result = add_numbers(10, 20)

print(result)
```

## 02. Return vs print()
- print() only prints.

```python
def add_numbers(a, b):
    # Print the result, but do not return it
    print(a + b)


result = add_numbers(10, 20)

print("Returned value:", result) # None will be printed because print() only prints the result. 
```

- return returns values.

```python
def add_numbers(a, b):
    # Return the result so it can be reused
    return a + b


result = add_numbers(10, 20)

print("Returned value:", result)
print("Result after multiplying by 2:", result * 2)
```

## 03. A return statement immediately stops function execution.

```python
def test_return():
    # This line will run
    print("Before return")

    return "Returned value"

    # This line will not run
    print("After return")


result = test_return()

print(result)
```

## 04. Functions without a return statement implicitly return None

```python
def greet_user(name):
    # Print a greeting message
    print(f"Hello, {name}!")


result = greet_user("World")

print("Returned value:", result)
```

## 06. Creating Multiple Functions
- The reason for creating multiple functions is to divide code based on specific roles.

```python
###########################################################################
# Example Scenarios: Processing a list of numbers via distinct functions: #
#    1. Calculate the total                                               #
#    2. Calculate the average                                             #
#    3. Find the maximum value                                            #
#    4. Format the result message                                         #
###########################################################################

def calculate_sum(numbers):
    # Calculate the sum of numbers in a list
    total = 0

    for number in numbers:
        total += number

    return total


def calculate_average(numbers):
    # Calculate the average of numbers in a list
    total = calculate_sum(numbers)
    average = total / len(numbers)

    return average


numbers = [10, 20, 30, 40, 50]

total_result = calculate_sum(numbers)
average_result = calculate_average(numbers)

print(f"Total: {total_result}")
print(f"Average: {average_result}")
```

## 07. Calling a Function Inside Another Function

```python
def calculate_sum(numbers):
    # Calculate the total sum
    total = 0

    for number in numbers:
        total += number

    return total


def create_summary(numbers):
    # Use another function to calculate the total
    total = calculate_sum(numbers)

    # Create a summary message
    message = f"The total sum is {total}."

    return message


numbers = [5, 10, 15]

summary = create_summary(numbers)

print(summary)
```

## 08. Returning multiple values
- Python returns multiple values as Tuple

```python
def calculate_sum_and_average(numbers):
    # Calculate the total sum
    total = sum(numbers)

    # Calculate the average
    average = total / len(numbers)

    return total, average


numbers = [10, 20, 30, 40, 50]

total_result, average_result = calculate_sum_and_average(numbers)

print(f"Total: {total_result}")
print(f"Average: {average_result}")

result = calculate_sum_and_average(numbers)

print(result)
print(type(result))
```

## Practice 6 — Create Sum and Average Functions
### Goal: Practice creating multiple functions.
##### Requirements
Create two functions:
1. calculate_sum(numbers)
   - Takes a list of numbers.
   - Returns the total sum.
2. calculate_average(numbers)
   - Takes a list of numbers.
   - Uses calculate_sum() inside the function.
   - Returns the average.<br>

Use this list:<br>

```python
def calculate_sum(numbers):
    sum_numbers = 0
    for number in numbers:
        sum_numbers += number
    return sum_numbers

def calculate_average(numbers):
    sum_num = calculate_sum(numbers)
    average_num = sum_num/len(numbers)
    return average_num 

numbers = [10, 20, 30, 40, 50]

sum_result = calculate_sum(numbers)
avg_result = calculate_average(numbers)

print(f"Sum numbers: {sum_result}")
print(f"Average numbers: {avg_result}")
```

### Practice 9 — Return Multiple Values

### Practice 10 — Small Function-Based Program
#### Goal: Practice building a small program using multiple functions.
##### Requirements
- Create the following functions:<br>
&nbsp;&nbsp;&nbsp;&nbsp;1. calculate_sum(numbers)<br>
&nbsp;&nbsp;&nbsp;&nbsp;2. calculate_average(numbers)<br>
&nbsp;&nbsp;&nbsp;&nbsp;3. check_score_level(average)<br>
&nbsp;&nbsp;&nbsp;&nbsp;4. create_summary_message(numbers)<br>

- Rules for check_score_level():<br>
&nbsp;&nbsp;&nbsp;&nbsp;90 or above: Excellent<br>
&nbsp;&nbsp;&nbsp;&nbsp;80 or above: Good<br>
&nbsp;&nbsp;&nbsp;&nbsp;70 or above: Average<br>
&nbsp;&nbsp;&nbsp;&nbsp;Below 70: Needs improvement<br>

- Use this list:<br>
&nbsp;&nbsp;&nbsp;&nbsp;scores = [85, 90, 78, 92, 88]

```python
def calculate_sum(numbers):
    sum_scores = 0
    for number in numbers:
        sum_scores += number
    return sum_scores
    
def calculate_average(numbers):
    sum_scores = calculate_sum(numbers)
    average_score = sum_scores/len(numbers)
    return average_score
    
def check_score_level(average):
    if average >= 90:
        return "Exellent"
    elif average >= 80:
        return "Good"
    elif average >= 70:
        return "Average"
    return "Needs improvement"
    
def create_summary_message(numbers):
    sum_num = calculate_sum(numbers)
    avg_num = calculate_average(numbers)
    check_score = check_score_level(avg_num)
    return f"Total: {sum_num}, Average: {avg_num:.2f}, Level: {check_score}"

scores = [85, 90, 78, 92, 88]

message = create_summary_message(scores)
print(message)
```
