# **Objective:** 
- Be able to group data by specific columns using pandas `groupby()`, and calculate summary statistics such as mean, sum, count, maximum, and minimum for each group.

# Today's Key Concepts
* **`groupby()`**: Splitting the DataFrame into groups based on specified column values
* **`mean()`**: Calculating the average value for each group
* **`sum()`**: Calculating the total sum of values for each group
* **`count()`**: Counting the number of non-null values in each group
* **`size()`**: Counting the total number of rows in each group (including null values)
* **`min()`**: Finding the minimum value within each group
* **`max()`**: Finding the maximum value within each group
* **`agg()`**: Applying multiple different aggregation functions to groups simultaneously
* **`reset_index()`**: Converting the grouped index back into regular DataFrame columns
* **Groupby with Multiple Columns**: Grouping data hierarchically by two or more columns
* **`sort_values()` After Groupby**: Sorting the aggregated results in ascending or descending order
* **Saving Grouped Results**: Exporting the final aggregated DataFrame to an external file

## groupby()
- groupby() allows you to split data into groups based on specific criteria, and then perform summary calculations for each group.

--- 

### Core Syntax: The `groupby()` Formula

```python
df.groupby("group_column")["target_column"].aggregation_function()
``` 

--- 

### Example: Calculating Departmental Averages

&nbsp;&nbsp;&nbsp;&nbsp;Let's assume we have the following employee dataset:

| Name | Department | Experience | Score | Status |
| :--- | :--- | :---: | :---: | :--- |
| Alice | Engineering | 5 | 88 | Active |
| Bob | Data | 3 | 92 | Active |
| Charlie | Engineering | 7 | 79 | Inactive |
| David | Cloud | 4 | 85 | Active |
| Eva | Data | 2 | 67 | Inactive |
| Frank | Security | 8 | 95 | Active |

#### &nbsp;&nbsp;&nbsp;&nbsp;Grouping by Department and Finding the Mean Score
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;If you want to view the **average score for each department**, you can group the data and calculate the mean like this:

```python
    # Group by Department and calculate the average of the Score column
    df_employees.groupby("Department")["Score"].mean()
```
#### &nbsp;&nbsp;&nbsp;&nbsp;Code Anatomy: Step-by-Step Execution

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Here is exactly how pandas evaluates the statement from left to right:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**`df_employees`** ➡️ The source **DataFrame** containing all raw data.<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**`.groupby("Department")`** ➡️ **Groups rows** based on unique categories within the "Department" column.<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**`["Score"]`** ➡️ **Selects the target column** to isolate the data we want to calculate.<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**`.mean()`** ➡️ **Computes the average** for each isolated group.
<br>
### Output
&nbsp;&nbsp;&nbsp;&nbsp;Department<br>
&nbsp;&nbsp;&nbsp;&nbsp;Cloud&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;85.0<br>
&nbsp;&nbsp;&nbsp;&nbsp;Data&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;79.5<br>
&nbsp;&nbsp;&nbsp;&nbsp;Engineering&nbsp;&nbsp;&nbsp;&nbsp;83.5<br>
&nbsp;&nbsp;&nbsp;&nbsp;Security&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;95.0<br>
&nbsp;&nbsp;&nbsp;&nbsp;Name: Score, dtype: float64<br><br>
### Data Flow Mechanism
&nbsp;&nbsp;&nbsp;&nbsp;Group rows by the "Department" column<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;↓<br>
&nbsp;&nbsp;&nbsp;&nbsp;Select the "Score" column within each Department<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;↓<br>
&nbsp;&nbsp;&nbsp;&nbsp;Calculate and output the Mean Score per Department<br>

### Creating a csv file for practice

````python
# Import modules
import os
import pandas as pd

# Create input and output folders
input_dir = "../assets/Week06/Day04/Input"
output_dir = "../assets/Week06/Day04/Output"

os.makedirs(input_dir, exist_ok=True)
os.makedirs(output_dir, exist_ok=True)

# Create sample employee performance data
employee_data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Henry"],
    "Department": ["Engineering", "Data", "Engineering", "Cloud", "Data", "Security", "Cloud", "Engineering"],
    "Experience": [5, 3, 7, 4, 2, 8, 6, 1],
    "Score": [88, 92, 79, 85, 67, 95, 90, 72],
    "Status": ["Active", "Active", "Inactive", "Active", "Inactive", "Active", "Active", "Inactive"],
    "Remote": [True, True, False, True, False, False, True, True]
}

# Create a DataFrame
df_employees = pd.DataFrame(employee_data)

# Save the DataFrame as a CSV file
df_employees.to_csv(f"{input_dir}/employee_performance.csv", index=False)

print("employee_performance.csv has been created.")
````

### Loading the created csv file

````python
# Read the employee performance CSV file
df_employees = pd.read_csv(f"{input_dir}/employee_performance.csv")

# Display the DataFrame
df_employees
````

## 01. Calculating the Average Score per Department

````python
department_avg_score = df_employees.groupby("Department")["Score"].mean()

department_avg_score
````

### Converting Groupby Results into a DataFrame
- Since `groupby()` often turns the grouping column into an index, you can use `.reset_index()` to convert it back into a standard DataFrame.

```Python
.groupby("group_column")["target_column"].aggregation_function().reset_index( name="new_column_name" )

````python
department_avg_score = df_employees.groupby("Department")["Score"].mean().reset_index( name="Mean" )

department_avg_score
````

## 02. Counting Employees by Department
- Use .size()

````python
department_employee_count = df_employees.groupby("Department").size().reset_index(name="Employee_count")

department_employee_count
````

## 03. Getting the Highest and Lowest Scores Together
- Use .max()

````python
department_max_score = df_employees.groupby("Department")["Score"].max().reset_index( name="Max score")

department_max_score
````

````python
department_min_score = df_employees.groupby("Department")["Score"].min().reset_index( name="Min score" )

department_min_score 
````

## 04. agg()
- Running Multiple Group Calculations at Once

---

### Syntax
1. Using a list => Target groups are in [] out of .agg()<br>
&nbsp;&nbsp;&nbsp;&nbsp;df_employees.groupby("Department")**["Score"]**.agg(<b>["mean", "max", "min"]</b>).reset_index()<br><br>
2. Using a dictionary => Target groups are in .agg() with dictionary form <br>
&nbsp;&nbsp;&nbsp;&nbsp;dept_summary = df_employees.groupby("Department").agg({<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;"Score": "mean",<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;"Experience": "sum"<br>
&nbsp;&nbsp;&nbsp;&nbsp;}).reset_index()

### 04-01: Using a list
- Running Multiple Group Calculations at Once

````python
department_score_summary = df_employees.groupby("Department")["Score"].agg(["mean","max","min","count"]).reset_index()

department_score_summary 
````

### 04-02: Using a dictionary
- Applying .agg() to Multiple Columns (looks like multi-level)

````python
department_summary = df_employees.groupby("Department").agg({
    "Score":["mean", "max", "min"],
    "Experience":["mean", "max"]
}).reset_index()

department_summary
````

- It also can be like this (not multi-level)<br>
&nbsp;&nbsp;&nbsp;&nbsp;df.**groupby**("<i>group_column</i>").**agg**(<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;title1 = (<i>"target_column</i>", "<i>aggregation_function()</i>"),<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;title2 = (<i>"target_column</i>", "<i>aggregation_function()</i>"),<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;title3 = (<i>"target_column</i>", "<i>aggregation_function()</i>"),<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;:<br>
&nbsp;&nbsp;&nbsp;&nbsp;).reset_index()

````python
# Create a clear grouped summary using named aggregation
department_summary = df_employees.groupby("Department").agg(
    Average_Score=("Score", "mean"),
    Max_Score=("Score", "max"),
    Min_Score=("Score", "min"),
    Employee_Count=("Name", "count"),
    Average_Experience=("Experience", "mean")
).reset_index()

department_summary
````

## 05. Sorting Groupby Results
- Use .sort_values

.**sort_values( by=**"<i>group_column</i>", **ascending=**<i>True/False</i> **)**

````python
department_summary_sorted = department_summary.sort_values(
    by="Average_Score",
    ascending=False
).reset_index(drop=True)

department_summary_sorted
````

## 06. Multi-column Groupby
- .grouby( [column_list] )

````python
department_status_summary = df_employees.groupby(["Department", "Status"]).size().reset_index(name="Employee_count")

department_status_summary
````

## 07. Grouping by a Boolean Column

````python
# Calculate average score by Remote status
remote_score_summary = df_employees.groupby("Remote")["Score"].mean().reset_index()

remote_score_summary
````

## 08. Saving groupby results

````python
department_summary_sorted.to_csv(f"{output_dir}/department_score_summary.csv", index=False)

print("department_score_summary.csv has been saved.")
````

````python
# Save a simple groupby summary as a text file
with open("../assets/Week06/Day04/Output/groupby_summary.txt", "w", encoding="utf-8") as file:
    file.write("GroupBy Summary\n")
    file.write("================\n")
    file.write(f"Total employees: {len(df_employees)}\n")
    file.write(f"Number of departments: {df_employees['Department'].nunique()}\n")
    file.write(f"Overall average score: {df_employees['Score'].mean():.2f}\n")
    file.write("\nTop department by average score:\n")
    file.write(f"{department_summary_sorted.iloc[0]['Department']}: {department_summary_sorted.iloc[0]['Average_Score']:.2f}\n")

print("groupby_summary.txt has been saved.")
````
