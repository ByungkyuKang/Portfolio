# **Objective:**
- Be able to add a new column to a pandas DataFrame and create calculated, conditional, or categorized columns using existing column values.

## Today's Key Concepts
* **Adding a New Column Directly**: Assigning a scalar value or an array to a new column label
* **Creating Calculated Columns Using Existing Columns**: Performing vector arithmetic between existing columns to derive a new one
* **Assigning Column Values Based on Conditions**: Populating a column with different values depending on conditional logic
* **Creating Boolean Columns**: Generating columns that store `True` or `False` based on logical expressions
* **Creating String Columns**: Creating columns containing text data or applying string methods to format text
* **Modifying Conditional Column Values with `.loc[]`**: Using label-based indexing to selectively update specific rows in a column
* **Saving a DataFrame with Newly Added Columns**: Persisting the updated DataFrame by exporting it to a file or assigning it back

````python
# pre seting
# Import modules
import os
import pandas as pd

# Create input and output folders
input_dir = "../assets/Week06/Day03/Input"
input_file = "employee_performance.csv"
output_dir = "../assets/Week06/Day03/Output"

os.makedirs(input_dir, exist_ok=True)
os.makedirs(output_dir, exist_ok=True)

# Create sample employee performance data
employee_data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Henry"],
    "Department": ["Engineering", "Data", "Engineering", "Cloud", "Data", "Security", "Cloud", "Engineering"],
    "Experience": [5, 3, 7, 4, 2, 8, 6, 1],
    "Score": [88, 92, 79, 85, 67, 95, 90, 72],
    "Status": ["Active", "Active", "Inactive", "Active", "Inactive", "Active", "Active", "Inactive"],
    "Remote": [True, True, False, True, False, False, True, True]
}

# Create a DataFrame
df_employees = pd.DataFrame(employee_data)

# Save the DataFrame as a CSV file
df_employees.to_csv(f"{input_dir}/{input_file}", index=False)

print("employee_performance.csv has been created.")

# Read the employee performance CSV file
df_employees = pd.read_csv("../assets/Week06/Day03/Input/employee_performance.csv")

# Display the DataFrame
df_employees
````

---
## 01. Adding a New Column with a Constant Value
- You can create a column that holds the same value for every single row. When you assign a scalar value, it is automatically populated across all rows.

````python
# Add a new column with the same value for all rows
df_employees["Company"] = "ABC Tech"

df_employees
````

---
## 02. Create a new calculated column using existing column values
- You can derive a new column by performing mathematical operations on existing column values. Pandas applies this vectorized arithmetic across all rows instantly.

````python
# Create a new column based on the Score column
df_employees["Score_Percentage"] = df_employees["Score"] / 100

df_employees
````

````python
# Convert years of experience to months
df_employees["Experience_Months"] = df_employees["Experience"] * 12

df_employees
````

---
## 03. Create a boolean column based on a conditional expression
- You can store the results of a conditional expression as a new column. This creates a Boolean column containing True or False values for each row based on the criteria.

````python
# Create a boolean column based on Score
df_employees["High_Performer"] = df_employees["Score"] >= 85

df_employees
````

- These boolean columns can be used directly as a boolean mask for filtering the DataFrame later

````python
# Filter only high performers
high_performers = df_employees[df_employees["High_Performer"]]

high_performers
````

---
## 04. Assign values conditionally

````python
###########################
# >= 90: Excellent        #
# < 90, >= 80: Good       #
# < 70, >= 70: Average    #
# < 70: Needs Improvement #
###########################

# Create a default score level column
df_employees["Score_Level"] = "Needs Improvement"

# Update Score_Level based on Score conditions
df_employees.loc[df_employees["Score"] >= 70, "Score_Level"] = "Average"
df_employees.loc[df_employees["Score"] >= 80, "Score_Level"] = "Good"
df_employees.loc[df_employees["Score"] >= 90, "Score_Level"] = "Excellent"

df_employees
````

---
## 05. Assign values based on a combination of multiple conditions

````python
# e.g.)
# Score >= 85 AND Status == Active

# Create a bonus eligibility column using multiple conditions
df_employees["Bonus_Eligible"] = (
    (df_employees["Score"] >= 85) & 
    (df_employees["Status"] == "Active")
)

df_employees
````

---
## 06. Concatenate multiple string columns to create a custom text feature

````python
# Create an employee label using Name and Department
df_employees["Employee_Label"] = df_employees["Name"] + " - " + df_employees["Department"]

df_employees
````

## 07. Use .copy() to add a new column after filtering
- If you filter rows and immediately try to add a new column, pandas will get confused and show a warning. To avoid this, append .copy() to your filtering code so you can work on a fresh, safe copy of the data

```python
# BAD CASE: Triggers SettingWithCopyWarning
engineering_staff = df_employees[df_employees["Department"] == "Engineering"]
engineering_staff["Bonus_Rate"] = 0.15  # Warning raised here!

# GOOD CASE: Clean, safe, and correct approach
engineering_staff = df_employees[df_employees["Department"] == "Engineering"].copy()
engineering_staff["Bonus_Rate"] = 0.15  # Works perfectly without any warnings
```

````python
# Filter high score employees and create a copy
high_score_employees = df_employees[df_employees["Score"] >= 85].copy()

# Add a new column to the copied DataFrame
high_score_employees["Review_Note"] = "High score review needed"

high_score_employees
````

---
## 08. Saving the Results with the New Columns

````python
# Save the DataFrame with new columns
df_employees.to_csv(
    "../assets/Week06/Day03/Output/employee_with_new_columns.csv",
    index=False
)

print("employee_with_new_columns.csv has been saved.")
````

````python
# Filter bonus eligible employees
high_score_bonus_employees = df_employees[df_employees["Bonus_Eligible"]].copy()

# Save filtered result
high_score_bonus_employees.to_csv(
    "../assets/Week06/Day03/Output/high_score_bonus_employees.csv",
    index=False
)

print("high_score_bonus_employees.csv has been saved.")
````
