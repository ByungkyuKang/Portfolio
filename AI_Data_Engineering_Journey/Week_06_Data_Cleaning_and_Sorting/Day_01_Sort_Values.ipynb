## Objective
- Be able to use pandas sort_values() to sort a DataFrame in ascending or descending order based on specific columns.

### Today's Key Concepts
- sort_values(): The core method used to sort a DataFrame by its values
- ascending=True: Sorting data in ascending order (lowest to highest, default)
- ascending=False: Sorting data in descending order (highest to lowest)
- Sorting by One Column: Sorting the entire DataFrame based on a single column's values
- Sorting by Multiple Columns: Sorting by multiple columns sequentially using a list (e.g., secondary sorting)
- reset_index(): Resetting the row index labels after sorting rearranges them
- inplace: Applying changes directly to the original DataFrame without creating a copy
- Saving Sorted Results: Assigning the sorted DataFrame back to a variable to persist the changes

## 01. sort_values()
- `sort_values()` is used to sort a DataFrame based on the values of a specific column.
- `sort_values()` returns a new DataFrame by default. `inplace=True` can be used to change the original DataFrame.

---

### Example: Student Score Dataset
Let's assume we have the following student dataset:

| Name | Age | Score | Department |
| :--- | :---: | :---: | :--- |
| Alice | 25 | 88 | Data |
| Bob | 30 | 92 | Engineering |
| Charlie | 35 | 79 | Data |
| David | 28 | 85 | Cloud |
| Eva | 22 | 67 | Engineering |
| Frank | 40 | 95 | Security |

#### 1) Sorting by Scores in Descending Order (Highest to Lowest)
&nbsp;&nbsp;&nbsp;&nbsp;If you want to view the students ranked from the highest score to the lowest, set `ascending=False`:
```python
df.sort_values(by="column", ascending=False)
```
#### 2) Sorting by Scores in Ascending Order (Lowest to Highest)
&nbsp;&nbsp;&nbsp;&nbsp;If you want to view the students ranked from the lowest score to the highest, set ascending=True (or omit it, as it is the default):
```Python
df.sort_values(by="column", ascending=True)
```

### Creating dataset to practice

````python
# Import modules
import os
import pandas as pd

# Basic environment settings
# Input
input_dir = "../assets/Week06/Day01/Input/"
os.makedirs(input_dir, exist_ok=True)

input_file = "students.csv"
input_file_path = f"{input_dir}{input_file}"


# Output
output_dir = "../assets/Week06/Day01/Output/"
os.makedirs(output_dir, exist_ok=True)



# Create sample student data
student_data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Age": [25, 30, 35, 28, 22, 40],
    "Score": [88, 92, 79, 85, 67, 95],
    "Department": ["Data", "Engineering", "Data", "Cloud", "Engineering", "Security"],
    "Passed": [True, True, False, True, False, True]
}

# Creating a DataFrame
df_students = pd.DataFrame(student_data)

# Creating a CSV file
df_students.to_csv(input_file_path, index=False)

print("students.csv has been created.")
````

````python
# Read the student CSV file
df_students = pd.read_csv(input_file_path)
print(df_students)
````

### Sorting in ascending numerical order

````python
asc_students = df_students.sort_values( by="Score", ascending=True )
asc_students
````

###  Sorting in descending numerical order

````python
desc_students = df_students.sort_values( by="Score", ascending=False )
desc_students
````

### Sorting in ascending alphabetical order

````python
asc_name_sorted = df_students.sort_values( by="Name", ascending=True )
asc_name_sorted
````

### Sorting in descending alphabetical order

````python
desc_name_sorted = df_students.sort_values( by="Name", ascending=False )
desc_name_sorted
````

### Sorting by Multiple Columns
- To sort a DataFrame by multiple columns, pass a list of column names to the by parameter. The DataFrame will be sorted sequentially based on the order of the columns in the list.

````python
df_multi_sort = df_students.sort_values( by=["Department", "Score"], ascending=True )
df_multi_sort
````

- `ascending=` can also have a list

````python
df_multi_sort = df_students.sort_values( by=["Department", "Score"], ascending=[True, False] )
df_multi_sort
````

## 02. Resetting the Index After Sorting
- Sorting data via sort_values() rearranges the rows, but keeps the old index unchanged. If you want to refresh the index so it goes in order from 0, 1, 2..., use `reset_index()`.
- By default, the old index gets added as a new column on the far left. If you want to prevent this and completely remove the old index, simply set `drop=True` inside the function.

````python
# By default, the old index gets added as a new column
sorted_df = df_students.sort_values(by="Score", ascending=False).reset_index()
sorted_df
````

````python
# Using drop=True to remove the old index
df_drop_index = df_students.sort_values(by="Score", ascending=False).reset_index(drop=True)
df_drop_index
````

## 03. inplace=True
- `inplace=True` modifies the original DataFrame directly without creating a new copy. When this parameter is set, the method updates the existing object and returns `None`.
``` Python
    df_students.sort_values(by="Score", ascending=False, inplace=True)
```

## 04. Save the result to a csv file

````python
df_sorted_by_score = df_students.sort_values( by="Score", ascending=False ).reset_index( drop=True )
df_sorted_by_score.to_csv(f"{output_dir}students_sorted_by_score.csv", index=False)

print("Sorted result has been saved.")
````
