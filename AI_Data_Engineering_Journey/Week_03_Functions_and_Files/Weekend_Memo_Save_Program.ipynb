## Key Features
1. Input: The user enters a memo title and content.
2. Format: Clean and format the input memo as a string.
3. Save: Save the formatted text to a .txt file.
4. Read & Output: Read the saved .txt file back and display its content.
5. Modularize: Divide these features into separate functions.

```python
# import modules
import os

# functions
def write_contents(title, memo, full_path_name):
    """ Functions to write """
    if os.path.exists(full_path_name):
        mode = "a"
    else:
        mode = "w"
        print("File does not exist. A new file will be created.")

    with open(full_path_name, mode, encoding="utf-8") as file:
        # Add a blank line before appending a new memo
        if mode == "a":
            file.write("\n")

        file.write(f"Title: {title}\n")
        file.write(f"Memo: {memo}\n")

file_path = "../assets/Weekend"
file_name = "memo.txt"
file_full_name = f"{file_path}/{file_name}"

os.makedirs(file_path, exist_ok=True)

# Getting a memo title
title_input = input("Enter the title: ")
memo_input = input("What do you want to keep? ")

write_contents(title_input, memo_input, file_full_name)
```
