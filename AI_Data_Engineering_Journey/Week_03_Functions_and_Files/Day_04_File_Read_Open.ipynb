## 01. Reading Files with with open()
- "r": read
- .read(): loads the whole file content into one continuous **string**.

```python
with open("../assets/Week05/Week03/Day03/day03_profile.txt", "r") as file:
    # Read the entire file content as one string
    content = file.read()

print(content)
print(type(content))
```

## 02. readline()
- The readline() method reads the file content one line at a time.
- Every time readline() is invoked, it reads the current line and moves to the next.

```python
with open("../assets/Week05/Week03/Day03/day03_profile.txt", "r") as file:
    # Read the first line
    first_line = file.readline()
    # Read the second line
    second_line = file.readline()

print(first_line)
print(second_line)
```

## 03. readlines()
- The readlines() method reads all lines of a file and returns them as a list.

```python
with open("../assets/Week05/Week03/Day03/day03_profile.txt", "r") as file:
    # Read all lines as a list
    lines = file.readlines()

print(lines)
print(type(lines))
```

- When a trailing \n is attached to the end of each line, you can use strip() to clean up the output for display.<br>

## 04. read(), readline(), readlines()
| Method | Description | Return Type |
|---|---|---|
| `read()` | Reads the entire file content | `str` |
| `readline()` | Reads one line at a time | `str` |
| `readlines()` | Reads all lines into a list | `list` |

## 05. Read with for loop

## 07. Preventing Errors When a File Is Missing
- Attempting to open a missing file triggers a FileNotFoundError.

```python
# An error occurs
with open("../assets/Week05/Week03/Day03/not_existing_file.txt", "r") as file:
    content = file.read()
```

- Use os.path.exists() to prevent the FileNotFoundError

```python
import os

file_path = "../assets/Week05/Week03/Day03/day03_profile.txt"

# Check if the file exists before reading
if os.path.exists(file_path):
    with open(file_path, "r") as file:
        content = file.read()

    print(content)
else:
    print("File does not exist.")
```

## 08. Reading and analyzing a file of numbers

```python
import os

os.mkdir("../assets/Day04/")

with open("../assets/Day04/Day04_numbers.txt", "w") as file:
    # Write one number per line
    file.write("10\n")
    file.write("20\n")
    file.write("30\n")
    file.write("40\n")
    file.write("50\n")
```

```python
numbers = []

with open("../assets/Day04/Day04_numbers.txt", "r") as file:
    # Read each line and convert it to an integer
    for line in file:
        numbers.append(int(line.strip()))

total = sum(numbers)
average = total / len(numbers)

print(f"Numbers: {numbers}")
print(f"Total: {total}")
print(f"Average: {average:.2f}")
```

### Practice 8 — Create and Read a Numbers File
#### Goal: Practice writing numbers to a file and reading them back as integers.
##### Requirements
Create a file called:<br>
&nbsp;&nbsp;&nbsp;&nbsp;../assets/Day04_numbers.txt<br><br>
Then read the file and calculate:<br>
&nbsp;&nbsp;&nbsp;&nbsp;total<br>
&nbsp;&nbsp;&nbsp;&nbsp;average

```python
read_path = "../assets/Day04/Day04_numbers.txt"

read_list = []
with open(read_path, "r") as file:
    for number in file.readlines():
        read_list.append(int(number.strip()))

print(read_list)
print(f"Total: {sum(read_list)}")
print(f"Average: {sum(read_list)/len(read_list):.2f}")
```

### Practice 10 — Create a File Reading Summary
#### Goal: Practice creating a small summary after reading a file.
##### Requirements
Read this file:<br>
&nbsp;&nbsp;&nbsp;&nbsp;../assets/Week05/Week03/Day03/day03_skills.txt<br><br>
Create a summary message that includes:<br>
&nbsp;&nbsp;&nbsp;&nbsp;Number of lines<br>
&nbsp;&nbsp;&nbsp;&nbsp;First line<br>

```python
read_path = "../assets/Day03/day03_skills.txt"

with open(read_path, "r") as file:
    # Read all lines into a list
    lines = file.readlines()

cleaned_lines = []

# Clean each line
for line in lines:
    cleaned_lines.append(line.strip())

line_count = len(cleaned_lines)
first_line = cleaned_lines[0]
last_line = cleaned_lines[-1]

print(f"Number of lines: {line_count}")
print(f"First line: {first_line}")
print(f"Last line: {last_line}")
```
