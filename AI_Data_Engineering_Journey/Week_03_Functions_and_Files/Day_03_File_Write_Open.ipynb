## 01. with open()
|Basic Structure|Meaning|
|---|---|
|<b>with open(</b>...<b>) as</b> <i>file</i>:|Open the file to use under the alias <i>`file`</i>, and let it close automatically after the block ends.|

### Modes ( "w"/"a"/"r" )
---

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<i>file</i>.write("<i>New content</i>")<br><br>
&nbsp;&nbsp;&nbsp;&nbsp;- "a" (Append): Add after existing content<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;with open("<i>file_name.txt</i>", **"<i>a</i>"**) as <i>file</i>:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<i>file</i>.write("<i>Content to add</i>")<br><br>

```python
# Basic structure
# - Create an output file and save a message
with open("../assets/Week03/Day03/Day_03_first_output.txt", "w") as file:
    file.write("Hello, world!\n")

# - Read the file create above
with open("../assets/Week03/Day03/Day_03_first_output.txt", "r") as file:
    content = file.read()

print(content)
```

```python
# - add a message to the existing output file
with open("../assets/Week03/Day03/Day_03_first_output.txt", "a") as file:
    file.write("Python!\n")

# - Read the file create above
with open("../assets/Week03/Day03/Day_03_first_output.txt", "r") as file:
    content = file.read()

print(content)    # Python! is added
```

## 02. os.makedirs() and exist_ok
- To avoid errors when saving files to missing directories, use os.makedirs() to build the path. By passing exist_ok=True, Python will safely skip creation without raising an exception if the folder is already there.<br><br>
&nbsp;&nbsp;&nbsp;&nbsp;os.makedirs(..., exist_ok=True)

```python
import os

# Create the outputs folder if it does not exist
os.makedirs("../assets/Week03/Day03", exist_ok=True)

with open("../assets/Week03/Day03/Day_03_first_output.txt", "w") as file:
    # Write a simple message to the output file
    file.write("This is my first text file created with Python.\n")
```

```python
## To confirm if it's created
with open("../assets/Week03/Day03/Day_03_first_output.txt", "r") as file:
    content = file.read()

print(content)
```

### Practice 5 — Write Number Summary to a File
#### Goal: Practice calculating values and saving the result to a text file.
##### Requirements
Create a list called numbers:<br>
&nbsp;&nbsp;&nbsp;&nbsp;numbers = [10, 20, 30, 40, 50]<br><br>
Calculate:<br>
&nbsp;&nbsp;&nbsp;&nbsp;total<br>
&nbsp;&nbsp;&nbsp;&nbsp;average<br><br>
Save the result into:<br>
&nbsp;&nbsp;&nbsp;&nbsp;../assets/Week03/Day03/day03_number_summary.txt<br><br>
Expected file content:<br>
&nbsp;&nbsp;&nbsp;&nbsp;Numbers: [10, 20, 30, 40, 50]<br>
&nbsp;&nbsp;&nbsp;&nbsp;Total: 150<br>
&nbsp;&nbsp;&nbsp;&nbsp;Average: 30.00

```python
# Answer
# Although sum() can calculate the total sum of a list, 
# I am creating a custom function for practice purposes.
def total_number(numbers):
    result = 0
    for number in numbers:
        result += number
    return result

def average_number(numbers):
    total_num = total_number(numbers)
    return total_num/len(numbers)
    
numbers = [10, 20, 30, 40, 50]

sum_num = total_number(numbers)
avg_num = average_number(numbers)

with open("../assets/Week03/Day03/day03_number_summary.txt", "w") as file:
    file.write(f"Numbers: {numbers}\n")
    file.write(f"Total: {sum_num}\n")
    file.write(f"Average: {avg_num:.2f}\n")
```

```python
# Checking the result
with open("../assets/Week03/Day03/day03_number_summary.txt", "r") as file:
    content = file.read()

print("Check Results:\n")
print(content)
```

### Practice 6 — Save Even and Odd Results
#### Goal: Practice combining loops, conditions, and file writing.
##### Requirements
Create a list called numbers:<br>
&nbsp;&nbsp;&nbsp;&nbsp;numbers = [3, 8, 15, 22, 31, 40]<br><br>
Save whether each number is even or odd into:<br>
&nbsp;&nbsp;&nbsp;&nbsp;../assets/Week03/Day03/day03_even_odd_results.txt<br><br>
Expected file content:<br>
&nbsp;&nbsp;&nbsp;&nbsp;3 is odd.<br>
&nbsp;&nbsp;&nbsp;&nbsp;8 is even.<br>
&nbsp;&nbsp;&nbsp;&nbsp;15 is odd.<br>
&nbsp;&nbsp;&nbsp;&nbsp;22 is even.<br>
&nbsp;&nbsp;&nbsp;&nbsp;31 is odd.<br>
&nbsp;&nbsp;&nbsp;&nbsp;40 is even.

```python
# Answer
numbers = [3, 8, 15, 22, 31, 40]

with open("../assets/Week03/Day03/day03_even_odd_results.txt", "w") as file:
    for number in numbers:
        if number%2==0:
            file.write(f"{number} is even.\n")
        else:
            file.write(f"{number} is odd.\n")
```

```python
# Checking the result
with open("../assets/Week03/Day03/day03_even_odd_results.txt", "r") as file:
    content = file.read()

print("Check Results:\n")
print(content)
```

### Practice 8 — Function and File Writing
#### Goal: Practice combining a function with file writing.
##### Requirements
Create a function called create_summary_message().<br><br>
The function should:<br>
&nbsp;&nbsp;&nbsp;&nbsp;1. Take day, topic, and status as input.<br>
&nbsp;&nbsp;&nbsp;&nbsp;2. Return a formatted summary message.<br>
&nbsp;&nbsp;&nbsp;&nbsp;3. Save the returned message into a text file.<br><br>
Save the result into:<br>
&nbsp;&nbsp;&nbsp;&nbsp;../assets/Week03/Day03/day03_summary_message.txt

```python
# Answer
def create_summary_message(day, topic, status):
    return f"Day: {day}\nTopic: {topic}\nStatus: {status}\n"

summary_message = create_summary_message(
    "Day 03",
    "File Write with open() and write()",
    "Completed"
)

with open("../assets/Week03/Day03/day03_summary_message.txt", "w") as file:
    file.write(summary_message)
```

### Practice 9 — Save Model Results
#### Goal: Practice saving structured result messages to a text file.
##### Requirements
Create the following dictionary:<br>
&nbsp;&nbsp;&nbsp;&nbsp;model_results = {<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;"Logistic Regression": 0.8234,<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;"Decision Tree": 0.7542,<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;"Random Forest": 0.8615<br>
&nbsp;&nbsp;&nbsp;&nbsp;}<br><br>
Save the model results into:<br>
&nbsp;&nbsp;&nbsp;&nbsp;../assets/Week03/Day03/day03_model_results.txt<br><br>
Expected file content:<br>
&nbsp;&nbsp;&nbsp;&nbsp;Model Results<br>
&nbsp;&nbsp;&nbsp;&nbsp;Logistic Regression: 82.34%<br>
&nbsp;&nbsp;&nbsp;&nbsp;Decision Tree: 75.42%<br>
&nbsp;&nbsp;&nbsp;&nbsp;Random Forest: 86.15%

```python
# Answer
model_results = {
    "Logistic Regression": 0.8234,
    "Decision Tree": 0.7542,
    "Random Forest": 0.8615
}

with open("../assets/Week03/Day03/day03_model_results.txt", "w") as file:
    file.write("Model Results\n")
    for key, value in model_results.items():
        file.write(f"{key}: {value*100:.2f}%\n")
```

```python
# Checking the result
with open("../assets/Week03/Day03/day03_model_results.txt", "r") as file:
    content = file.read()

print("Check Results:\n")
print(content)
```
