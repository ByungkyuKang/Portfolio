## 01. Functions
- A named, reusable block of code for executing repetitive tasks.

```python
# Without a function. 
print("Without a function:")

a = 10
b = 20
result = a + b
print(result)

a = 30
b = 40
result = a + b
print(result)

a = 100
b = 200
result = a + b
print(result)

# With a function - Reuse the function for repetitive tasks
print("With a function:")
def add_numbers(a, b):
    # Add two numbers and return the result
    return a + b

print(add_numbers(10, 20))
print(add_numbers(30, 40))
print(add_numbers(100, 200))
```

## 02. Basic Structure of a Function
def function_name(parameter):<br>
&nbsp;&nbsp;&nbsp;&nbsp;# Code block<br>
&nbsp;&nbsp;&nbsp;&nbsp;return result<br>

```python
# Example
# def          → Defines a function
# greet        → The function's name
# name         → Parameter
# return       → Returns the result
def greet(name):
    # Create a greeting message
    return f"Hello, {name}!"


message = greet("World")
print(message)
```

## 03. Parameter vs. Argument
- **Parameters**: The names of input variables used when defining a function.
- **Argumaents**: The actual values passed into a function when it is called.

```python
# Parameter = name
def greet(name):
    return f"Hello, {name}!"

# Argument = World
message = greet("World")
print(message)
```

## 04. Return
- Returning the processed result of a function back to the caller.

```python
def add_numbers(a, b):
    # Return the sum of two numbers
    return a + b

result = add_numbers(10, 20)

print(result)
```

## 05. print() vs. return
- **print()**: displays data on the screen.
- **return**: Returning the processed result of a function back to the caller.

```python
# print()
print("print():")
def add_numbers(a, b):
    # Print the result directly
    print(a + b)

result = add_numbers(10, 20)

print("Returned value:", result) # Nothing to return, so it returns None.

# Return
print("\nreturn")
def add_numbers(a, b):
    # Return the result so it can be reused later
    return a + b


result = add_numbers(10, 20)

print("Returned value:", result)
```

### Practice 10 — Function Review Program
#### Goal: Practice combining multiple basic concepts in one small program.
##### Requirements
Create the following functions:
1. calculate_sum(numbers)
2. calculate_average(numbers)
3. check_even_odd(number)

Then test each function with sample values.

```python
def calculate_sum(numbers):
    sum_num = 0
    for number in numbers:
        sum_num += number
    return sum_num
    
def calculate_average(numbers):
    sum_num = 0
    for number in numbers:
        sum_num += number
    return sum_num/len(numbers)
    
def check_even_odd(numbers):
    result_list = []
    for number in numbers:
        if number%2==0:
            result_message = f"{number} is even."
            result_list.append(result_message)
        else:
            result_message = f"{number} is odd."
            result_list.append(result_message)
    return result_list

numbers = [10, 23, 31, 42, 57]

total_result = calculate_sum(numbers)
avg_result = calculate_average(numbers)
odd_even_result = check_even_odd(numbers)

print(f"Numbers: {numbers}")
print(f"Total: {total_result}")
print(f"Average: {avg_result:.2f}")

for mgs in odd_even_result:
    print(mgs)
```
