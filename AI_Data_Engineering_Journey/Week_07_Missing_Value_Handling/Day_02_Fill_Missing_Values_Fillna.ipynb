## Setting

````python
import os
import pandas as pd
from pathlib import Path

# Create input and output folders
input_dir = "../assets/Week07/Day02/Input"
output_dir = "../assets/Week07/Day02/Output"

os.makedirs(input_dir, exist_ok=True)
os.makedirs(output_dir, exist_ok=True)

# Define input file path
input_file_path = f"{input_dir}/employee_data_with_missing_values.csv"

path = Path(input_file_path)

# Create sample employee data with missing values
employee_data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Henry"],
    "Department": ["Engineering", "Data", None, "Cloud", "Data", "Security", None, "Engineering"],
    "Experience": [5, 3, 7, None, 2, 8, 6, 1],
    "Score": [88, 92, None, 85, 67, 95, 90, None],
    "Status": ["Active", "Active", "Inactive", "Active", None, "Active", "Active", "Inactive"],
    "Remote": [True, True, False, True, None, False, True, True]
}
# Create DataFrame
df_employees = pd.DataFrame( employee_data )

# Save as CSV
df_employees.to_csv(input_file_path, index=False)


if path.exists():
    print("employee_data_with_missing_values.csv has been created.")
else:
    print("File not created.")
````

````python
# Read the employee CSV file
df_employees = pd.read_csv(
    "../assets/Week07/Day02/Input/employee_data_with_missing_values.csv"
)

# Display the DataFrame
df_employees
````

## <u>fillna()</u>
- fillna() is a pandas function used to fill missing values, or NaN values, with specified alternative values.

&nbsp;&nbsp;&nbsp;&nbsp;e.g.) Replace NaN to "Unknown"<br>
``` Python
df_employees["Department"] = df_employees["Department"].fillna("Unknown")
```

- Checking Missing Values

````python
df_employees.isnull().sum()
````

### 01. Replacing NaN in a specific column with a fixed value
- NaN in Department -> Unknown

````python
df_department_filled = df_employees.copy()
df_department_filled["Department"] = df_department_filled["Department"].fillna("Unknown")

df_department_filled
````

````python
df_department_filled.isnull().sum()
````

### 02. Replacing NaN in a numeric column with the mean value.

````python
df_score_mean_filled = df_employees.copy()
average_score = df_score_mean_filled["Score"].mean()

df_score_mean_filled = df_score_mean_filled["Score"].fillna(round(average_score, 2))
df_score_mean_filled
````

### 03. Replacing NaN in a numeric column with the memian value.

````python
df_experience_median_filled = df_employees.copy()
median_experiment = df_experience_median_filled["Experience"].median()
df_experience_median_filled = df_experience_median_filled["Experience"].fillna(median_experiment)

df_experience_median_filled
````

### 04. Imputing a String Column with the Mode (.mode())
- Since string columns cannot be averaged, you can fill missing values with the most frequent value using the `.mode()` function.
- `.mode()` returns a series, so [0] is used to retrieve the most common value

````python
df_status_mode_filled = df_employees.copy()

most_common_status = df_status_mode_filled["Status"].mode()[0]

print(f"The most common value is {most_common_status}.")

df_status_mode_filled["Status"] = df_status_mode_filled["Status"].fillna(most_common_status)

df_status_mode_filled
````

### 05. Filling a Boolean Column

````python
df_remote_filled = df_employees.copy()

# Replacing NaN to False. This is just a practice.
df_remote_filled["Remote"] = df_remote_filled["Remote"].astype(bool).fillna(False)

df_remote_filled
````

### 06. Fill up multiple columns at once: Use a dictionary
- Department → "Unknown"
- Experience → median
- Score → mean
- Status → mode
- Remote → False

````python
df_filled = df_employees.copy()

df_filled["Remote"] = df_filled["Remote"].astype(bool)

# Calculate fill values
experience_median = df_filled["Experience"].median()
score_mean = round(df_filled["Score"].mean(), 2)
status_mode = df_filled["Status"].mode()[0]

# Fill missing values by column
df_filled = df_filled.fillna(
    {
        "Department": "Unknown",
        "Experience": experience_median,
        "Score": score_mean,
        "Status": status_mode,
        "Remote": False
    }
)

df_filled
````

### 07. Check if all NaNs are replaced
- Use `.isnull().sum()`

````python
df_filled.isnull().sum()
````

````python
# Create missing value counts before fillna
missing_before = df_employees.isnull().sum()

# Create missing value counts after fillna
missing_after = df_filled.isnull().sum()

# Create fillna summary DataFrame
fillna_summary = pd.DataFrame({
    "Column": df_employees.columns,
    "Missing_Before": missing_before.values,
    "Missing_After": missing_after.values
})

fillna_summary
````

````python
# Define output file paths
filled_file_path = f"{output_dir}/employee_data_filled.csv"
fillna_summary_file_path = f"{output_dir}/fillna_summary.csv"

# Save filled DataFrame
df_filled.to_csv(filled_file_path, index=False)
filled_csv_file_path = Path(filled_file_path)

#Save fillna summary
fillna_summary.to_csv(fillna_summary_file_path, index=False)
fillna_summary_csv_path  = Path(fillna_summary_file_path)

if filled_csv_file_path.exists() and fillna_summary_csv_path:
    print("Filled data and fillna summary files have been saved.")
else:
    print("Files not created.")
````

### Practice Dataset — Customer Orders

````python
# Create practice dataset with missing values
order_data = {
    "Order_ID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008],
    "Customer": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Henry"],
    "Region": ["East", "West", None, "South", "East", None, "West", "South"],
    "Order_Amount": [250, None, 180, 320, None, 500, 410, None],
    "Discount": [10, 5, None, 15, 0, None, 20, 10],
    "Payment_Method": ["Credit Card", None, "PayPal", "Credit Card", "Cash", "Credit Card", None, "PayPal"],
    "Delivered": [True, True, False, None, True, False, True, None]
}

df_orders = pd.DataFrame(order_data)

df_orders
````

### Practice 2 — Fill Region with "Unknown"

##### <u>Requirements</u>
&nbsp;&nbsp;&nbsp;&nbsp;Fill missing values in Region with:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Unknown<br><br>

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;df_orders_filled<br><br>

````python
# Fill missing values in Regions with "Unknown"
df_orders_filled = df_orders.copy()

df_orders_filled["Region"] = df_orders_filled["Region"].fillna("Unknown")
df_orders_filled
````

### Practice 3 — Fill Order_Amount with the Median
#### Goal: Practice filling a numeric column with a median value.
##### <u>Requirements</u>
&nbsp;&nbsp;&nbsp;&nbsp;Calculate the median of Order_Amount.<br>
&nbsp;&nbsp;&nbsp;&nbsp;Fill missing values in Order_Amount with that median.

````python
df_order_amount_filled = df_orders.copy()

order_amount_median = df_order_amount_filled["Order_Amount"].median()
print(f"The median value: {order_amount_median}")

df_order_amount_filled["Order_Amount"] = df_order_amount_filled["Order_Amount"].fillna(order_amount_median)

df_order_amount_filled
````

### Practice 4 — Fill Discount with 0
#### Goal: Practice choosing a value based on column meaning.
##### <u>Requirements</u>
&nbsp;&nbsp;&nbsp;&nbsp;Fill missing values in Discount with 0.<br>

&nbsp;&nbsp;&nbsp;&nbsp;Reason:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;If Discount is missing, assume no discount was applied.

````python
df_discount_filled = df_orders.copy()

df_discount_filled["Discount"] = df_discount_filled["Discount"].fillna(0)
df_discount_filled
````

### Practice 5 — Fill Payment_Method with the Most Frequent Value
#### Goal: Practice filling a categorical column with the mode.
##### <u>Requirements</u>
&nbsp;&nbsp;&nbsp;&nbsp;Find the most common value in Payment_Method.<br>
&nbsp;&nbsp;&nbsp;&nbsp;Fill missing values in Payment_Method with that value.

````python
df_pay_method_filled = df_orders.copy()

most_used_pay_method = df_pay_method_filled["Payment_Method"].mode()[0]

df_pay_method_filled["Payment_Method"] = df_pay_method_filled["Payment_Method"].fillna(most_used_pay_method)

df_pay_method_filled
````

### Practice 6 — Fill Delivered with `False`
#### Goal: Practice filling a boolean column.
##### <u>Requirements</u>
&nbsp;&nbsp;&nbsp;&nbsp;Fill missing values in Delivered with `False`<br><br>
&nbsp;&nbsp;&nbsp;&nbsp;Reason:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;If delivery status is missing, assume it has not been delivered yet.

````python
df_delivered_filled = df_orders.copy()

df_delivered_filled["Delivered"] = df_delivered_filled["Delivered"].astype(bool).fillna(False)

df_delivered_filled
````

### Practice 7 — Create a Before/After Missing Value Summary
#### Goal: Practice comparing missing values before and after filling.
##### <u>Requirements</u>
&nbsp;&nbsp;&nbsp;&nbsp;Create a DataFrame called:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;order_fillna_summary<br><br>
&nbsp;&nbsp;&nbsp;&nbsp;It should have these columns:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Column<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Missing_Before<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Missing_After<br>

````python
df_orders_copy = df_orders.copy()
df_orders_copy["Delivered"] = df_orders_copy["Delivered"].astype(bool)

order_amount_median = df_orders_copy["Order_Amount"].median()
most_payment_method = df_orders_copy["Payment_Method"].mode()[0]

missing_before = df_orders_copy.isnull().sum()

missing_after = df_orders_copy.fillna({
    "Region" : "Unknown",
    "Order_Amount" : order_amount_median,
    "Discount" : 0,
    "Payment_Method" : most_payment_method,
    "Delivered" : False
}).isnull().sum()

order_fillna_summary = pd.DataFrame( {
    "Columns" : df_orders_copy.columns,
    "Missing_Before" : missing_before.values,
    "Missing_After" : missing_after.values
})
order_fillna_summary
````

### Practice 8 — Verify That All Missing Values Are Filled
#### Goal: Confirm that there are no missing values left.
##### <u>Requirements</u>
&nbsp;&nbsp;&nbsp;&nbsp;Print the total number of missing values after filling.<br>
&nbsp;&nbsp;&nbsp;&nbsp;Expected result: 0

````python
print(missing_after.isnull().sum().sum())
````

### Practice 9 — Add a New Column After Filling
#### Goal: Combine Week 6 Day 3 knowledge with `fillna()`.
##### <u>Requirements</u>
&nbsp;&nbsp;&nbsp;&nbsp;After filling missing values, create a new column called:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Final_Amount<br><br>
&nbsp;&nbsp;&nbsp;&nbsp;Formula:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Final_Amount = Order_Amount - Discount

````python
add_final_amount = df_orders_copy.fillna({
    "Region" : "Unknown",
    "Order_Amount" : order_amount_median,
    "Discount" : 0,
    "Payment_Method" : most_payment_method,
    "Delivered" : False
})

add_final_amount["Final_Amount"] = add_final_amount["Order_Amount"] - add_final_amount["Discount"]
add_final_amount
````
