## Required Features
1. User can enter the start table and end table.
2. Program generates multiplication tables for the selected range.
3. Program formats the multiplication table result clearly.
4. Program creates an output folder if it does not exist.
5. Program saves the result as a .txt file.
6. Program reads the saved file again to confirm the result.
7. Code is organized into functions.
8. Notebook includes explanation, code, output, and summary.

```python
# import modules
import os

# functions
# - Read the saved table (multiplication_table.txt)
def read_saved_table(save_file_dir):
    with open(save_file_dir, "r") as file:
        loaded_content = file.readlines()
    # print(loaded_content)
    for line in loaded_content:
        if line == '\n':
            print()
        else:
            if line.strip().endswith("|"):
                print(f"{line.strip()}")
            else:
                print(f"  {line.strip()}")
                
# - Save the table to a text file (multiplication_table.txt)
def save_table_file(created_table_list, save_file_dir):
    with open(save_file_dir, "w") as file:
        for value in created_table_list:
            file.write(f"{value}\n")
    
# - Create the table
def create_table(start, end, save_file_dir):
    if start <= end:
        table_list = []
        for table_number in range(start, end+1):
            times_table_title = f"| {table_number:02d} Times Table |"
            table_list.append(times_table_title)
            print(times_table_title)
            
            for number in range(1, 10):
                each_line = f"  {table_number:02d} * {number:02d} = {int(table_number)*int(number):02d}"
                table_list.append(each_line)
                print(each_line)
                
            table_list.append("")
            print("")
        save_table_file(table_list, save_file_dir)
    else:
        print(f"End table({end}) cannot be smaller than start table({start}).")

save_file_path = "../assets/Week04/mini_project"
save_file_name = "multiplication_tables.txt"
save_file_full = f"{save_file_path}/{save_file_name}"

os.makedirs(save_file_path, exist_ok=True)

while True:
    if os.path.exists(save_file_full):
        load_file = input("Do you want to load the table file?(Y/N) ")
        if load_file.lower() == 'y':
            read_saved_table(save_file_full)
            break
        else:
            start_table = int(input("Enter Start Table: "))
            end_table = int(input("Enter End Table: "))
            
            create_table(start_table, end_table, save_file_full)
    else:
        start_table = int(input("Enter Start Table: "))
        end_table = int(input("Enter End Table: "))
        
        create_table(start_table, end_table, save_file_full)

    try_again = input("Try again?(Y/N) ")
    if try_again.lower() == 'n':
        break
        
```
