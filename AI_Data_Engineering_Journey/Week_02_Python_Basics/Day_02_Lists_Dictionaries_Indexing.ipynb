## 01. List - []

```python
numbers = [10, 20, 30, 40, 50] # number
names = ["Alice", "Bob", "Charlie"] # string
mixed_values = ["Python", 100, 3.14, True] # diffrent types can be in a list

print(numbers)
print(names)
print(mixed_values)
```

#### Getting the number of emements in a list
- Len()

```python
skills = ["Python", "SQL", "AWS"]

print(len(skills))
```

#### Adding values to a list
- .append()

```python
skills = ["Python", "SQL", "AWS"]

print(f"Original: {skills}")

skills.append("Machine Learning")

print(f"Added   : {skills}")
```

#### Modifying lists
- list[index] = new_value

```python
# index     [0]      [1]    [2]
skills = ["Python", "SQL", "AWS"]

skills[1] = "Oracle SQL" # SQL -> Oracle SQL

print(skills)
```

#### Deleting values in a list
- by values: .remove( 'value' )
- by indexes: .pop( 'index' )

```python
# .remove()
skills = ["Python", "SQL", "AWS"]

skills.remove("SQL") # This removes the first 'SQL' in the list

print(skills)

# .pop()
skills = ["Python", "SQL", "AWS"]

removed_skill = skills.pop(1) # .pop saves the popped value

print(skills)
print(removed_skill)
```

## 02. Indexing
- Indexing is a way to get a value at a specific position from a list or a string.\
- Index start from 0

```python
# Index:      0        1       2
# Value:   Python     SQL     AWS

skills = ["Python", "SQL", "AWS"]

print(skills[0], skills[1], skills[2], sep=', ')
```

#### Negative indexing: counting from the end

```python
# Index:     -3        -2      -1
# Value:   Python     SQL     AWS

skills = ["Python", "SQL", "AWS"]

print(skills[-1], skills[-2], skills[-3], sep=', ')
```

- Indexing is supported for both lists and strings.

```python
language = "Python"

print(language[0])
print(language[1])
print(language[-1])
```

#### slicing: list[ first_index : end_index+1 ]
- Omitting the first index: from the first
- Omitting the end index: to the last
- Negative index can be used

```python
skills = ["Python", "SQL", "AWS", "Machine Learning", "Pandas"]

print(skills[:3])   # from the first to index 2
print(skills[2:])   # from index 2 to the end
print(skills[-2:])  # The last two elements
```

- Slicing is supported for both lists and strings.

```python
language = "Python"

print(language[0:3])
print(language[:2])
print(language[2:])
print(language[-2:])
```

- Retrieving values from a dictionary: a_dictionary[ key ]

```python
person = {
    "name": "Byungkyu",
    "age": 35,
    "role": "Software Developer"
}

print(person["name"])
print(person["age"])
print(person["role"])
```

- Adding values to a dictionary: a_dictionary[ new_key ]=value

```python
person = {
    "name": "Byungkyu",
    "age": 35
}

print(person)

person["role"] = "Software Developer"

print(person)
```

- Modifying a value in a dictionary: a_dictionary[ existing_key ]=value

```python
person = {
    "name": "Byungkyu",
    "age": 35,
    "role": "Software Developer"
}

print(person)

person["age"] = 20

print(person)
```

- .get() - Printing a message when a key doesn't exist

```python
person = {
    "name": "Byungkyu",
    "age": 35
}

print(person["role"]) # An error occurs
```

```python
print(person.get("role", "role doesn't exist")) # The error is handled
```

- Dictionary keys, values, items<br>
.keys(): returns a view of keys<br>
.values(): returns a view of values<br>
.items(): returns a view of key-value pair tuples<br>

```python
person = {
    "name": "Byungkyu",
    "age": 35,
    "role": "Software Developer"
}

print(person.keys())
print(person.values())
print(person.items())
```
