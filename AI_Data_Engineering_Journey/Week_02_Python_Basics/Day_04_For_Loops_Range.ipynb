## 01. For loop

&nbsp;&nbsp;&nbsp;&nbsp;<b>for</b> <i>item</i> <b>in</b> <i>iterable</i><b>:</b><br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<i>code_to_repeat</i>

```python
skills = ["Python", "SQL", "Pandas", "Machine Learning"]

# Print each skill in the list
for skill in skills:
    print(skill)
```

## 02. for loop + lists

```python
model_names = ["Logistic Regression", "Decision Tree", "Random Forest"]

# Print each model name
for model in model_names:
    print(f"Model: {model}")
```

## 03. for loop + string

```python
language = "Python"

# Print each character in the string
for char in language:
    print(char)
```

## 04. for loop + dictionary
- By default, iterating over a dictionary retrieves its keys one by one. ( = dictionary.**keys()** )

```python
profile = {
    "name": "Byungkyu",
    "role": "Software Developer",
    "age": 35
}

# Loop through dictionary keys
for key in profile:
    print(key)

print("")
# Same as above. ( .keys() )
for key in profile.keys():
    print(key)
```

- Loop through dictionary key-value pairs: dictionary<b>.items()</b>

```python
profile = {
    "name": "Byungkyu",
    "role": "Software Developer",
    "age": 35
}

# Loop through dictionary key-value pairs
for key, value in profile.items():
    print(f"{key}: {value}")
```

- Loop through dictionary only values: dictionary<b>.values()</b>

```python
profile = {
    "name": "Byungkyu",
    "role": "Software Developer",
    "age": 35
}

# Loop through dictionary key-value pairs
for value in profile.values():
    print(value)
```

## 05. range()
- to generate a sequence of numbers based on specified start, stop, and step values.

```python
# Print numbers from 0 to 4
for number in range(5): # 5 isn't included
    print(number)
```

## 06. range(start, stop)
- start number can be specified

```python
# Print numbers from 1 to 5
for number in range(1, 6):
    print(number)
```

## 07. range(start, stop, step)
- The increment interval can also be specified.

```python
# Print even numbers from 2 to 10
for number in range(2, 11, 2):
    print(number)
```

## 08. range() + for loop

```python
# Repeat the same message 3 times
for i in range(3):
    print("Practice Python basics.")
```

## 09. Use list index and range() together

```python
skills = ["Python", "SQL", "Pandas"]

# To handle both the index and value of a list simultaneously, range(len(list)) can be used.
for i in range(len(skills)):
    print(f"Index {i}: {skills[i]}")
```

- **enumerate()**: to get both index and value

```python
# enumerate()
skills = ["Python", "SQL", "Pandas"]

# Use enumerate to get both index and value
for index, skill in enumerate(skills):
    print(f"Index {index}: {skill}")
```

## 10. Accumulating Calculations in Loops

```python
scores = [0.82, 0.75, 0.86, 0.79]

total_score = 0

# Add each score to total_score
for score in scores:
    total_score = total_score + score

average_score = total_score / len(scores)

print(f"Total score: {total_score:.2f}")
print(f"Average score: {average_score:.2f}")
```

### Practice 12 — Find High Performing Models
#### Goal: Practice using a for loop with a condition.
##### Requirements

Create the following dictionary:<br><br>
