# 02. Numbers

```python
# 2 main types - int and float
integer_number = 10
decimal_number = 10.5
```

```python
print(integer_number)
print(decimal_number)
```

```python
# 기본 연산 
# + : addition
# - : subtraction
# * : multiplication
# / : division
# // : floor division
# % : remainder
# ** : exponent
a = 10
b = 3

print(a + b)  # addition
print(a - b)  # subtraction
print(a * b)  # multiplication
print(a / b)  # division
print(a // b) # floor division
print(a % b)  # remainder
print(a ** b) # exponent
```

# 03. Strings

```python
# Both '' and ""  can be used
language = 'Python'
tool = "Jupyter Notebook"

print(language)
print(tool)
```

```python
# Connecting strings
first_name = "Byungkyu"
last_name = "Kang"

full_name = first_name + " " + last_name
print(full_name)
print(first_name, last_name)
```

```python
# f-string
name = "Byungkyu"
age = 35

print(f"My name is {name}, and I'm {age} years old.")
```

```python
# methods for string (lower(), upper(), strip(), replace(), title())
text = "  Word methods  "

print(text.lower())
print(text.upper())
print(text.strip())
print(text.replace("Words", "String"))
print(text.title())
```

```python
# Can be used together
column_name = " Customer Name "
clean_column_name = column_name.strip().lower().replace(" ", "_")

print(clean_column_name)
```

### Practice 2 - Numeric Calculation
- Calculate the study time with the data below.

     daily_study_hours = 2.5<br>
     study_days_per_week = 6<br>
     weeks = 48

```python
daily_study_hours = 2.5
study_days_per_week = 6
weeks = 48

weekly_study_hours = daily_study_hours * study_days_per_week
total_study_hours = weekly_study_hours * weeks

print(f"Weekly study hours: {weekly_study_hours}")
print(f"Total study hours for {weeks} weeks: {total_study_hours}")
```
