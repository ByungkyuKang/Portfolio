## 01. Comparison Operators

| Operator | Meaning                  | Example   |
| -------- | ------------------------ | --------- |
| `==`     | Equal to                 | `x == 10` |
| `!=`     | Not equal to             | `x != 10` |
| `>`      | Greater than             | `x > 10`  |
| `<`      | Less than                | `x < 10`  |
| `>=`     | Greater than or equal to | `x >= 10` |
| `<=`     | Less than or equal to    | `x <= 10` |

### Returns True or False

```python
score = 85

# Check if the score is equal to 85
print(score == 85)

# Check if the score is not equal to 90
print(score != 90)

# Check if the score is greater than 80
print(score > 80)

# Check if the score is less than 70
print(score < 70)

# Check if the score is greater than or equal to 85
print(score >= 85)

# Check if the score is less than or equal to 100
print(score <= 100)
```

## 02. = vs ==
- =: to store values
- ==: to compare

```python
score = 85  # storing 85 in score
print(score)
score == 85 # Comparing the value in score and 85. It returns True when they are the same and Flase when they are different
```

## 03. if
- Runs if block when only it's True<br><br>
&nbsp;&nbsp;&nbsp;&nbsp;if condition:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;code_to_run<br>

```python
score = 85

# Run this block only if the score is greater than or equal to 80
if score >= 80:
    print("Good score")
```

```python
score = 65

# This block will not run because the condition is False
if score >= 80:
    print("Good score")
```

## 04. Indentation
- In Python, the if statement uses indentation to define code blocks instead of curly braces ({})

```python
# Correct example
score = 85

# The indented line belongs to the if block
if score >= 80:
    print("Good score")
```

```python
# Error (IndentationError)
score = 85

if score >= 80:
print("Good score")
```

## 05. if-else
- The if-else statement executes different code blocks depending on whether the condition is true or false.<br><br>
&nbsp;&nbsp;&nbsp;&nbsp;if <i>condition</i>:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<i>code_if_true</i><br>
&nbsp;&nbsp;&nbsp;&nbsp;else:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<i>code_if_false</i><br>

```python
score = 75

# Run one block if the condition is True, otherwise run the else block
if score >= 80:
    print("Pass with a good score")
else:
    print("Needs improvement")
```

## 06. if-elif-else
- for multiple conditions
- The if-elif-else statement evaluates conditions from top to bottom, executing **only the first block that evaluates to true**

```python
score = 85

# Classify the score into a grade
if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: Needs improvement")
```

## 07. Importance of condition order
- Placing conditions incorrectly can lead to unintended or incorrect results.

```python
# An example of incorrect condition orders
score = 95

# This condition order is not ideal
if score >= 70:
    print("Grade: C or above")
elif score >= 90:
    print("Grade: A")
```

## 09. and, or, not
- **and**: True when both values are True
- **or**: True when a value is True
- **not**: True when Fase, False when True

```python
# and
score = 85
attendance = 90

# Both conditions must be True
if score >= 80 and attendance >= 80:
    print("Passed")
else:
    print("Failed")

# When one is False
attendance = 50

if score >= 80 and attendance >= 80:
    print("Passed")
else:
    print("Failed")
```

```python
# or
file_type = "json"

# At least one condition must be True
if file_type == "csv" or file_type == "json":
    print("Supported file format")
else:
    print("Unsupported file format")
```

```python
# not
is_completed = False

# Run this block if the task is not completed
if not is_completed:
    print("Task is still in progress")
```

## 10. Boolean and if
