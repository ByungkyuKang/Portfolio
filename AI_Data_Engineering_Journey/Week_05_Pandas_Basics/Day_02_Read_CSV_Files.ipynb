## 03. Creating a CSV file - to_csv()
- pd<b>.to_csv(</b><i>"csv_save_path"<i><b>)</b>

```python
import pandas as pd
import os

# path and file
output_dir = "../assets/Week05/Day02/Output/"
output_file = "students.csv"
full_path = f"{output_dir}{output_file}"

# Create a folder for outputs
os.makedirs(output_dir, exist_ok=True)

# Create sample student data
student_data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 35, 28],
    "Score": [88, 92, 79, 85]
}

# Create a DataFrame
df_students = pd.DataFrame(student_data)

# Store the DataFrame as a CSV file
# Setting "index=False" excludes the row index when saving the DataFrame to a CSV.
df_students.to_csv(full_path, index=False)

if os.path.exists(full_path):
    print("students.csv has been created.")
else:
    print("Error - file not created.")
```

## 05. Frequently used options
- sep
- usecols
- nrows

### `sep`: By default, CSV files use a comma (,) separator, so setting sep="," is optional. However, if a file is delimited by other characters like ; or |, you can easily handle it by setting the `sep` option to the corresponding character.

- pd.read_csv("<i>path/file.csv</i>", **sep="**,**"**)

### `usecols`: Used when you want only to read specific columns

```python
df_selected = pd.read_csv(
    "../assets/Week05/Day02/Input/students.csv",
    usecols=["Name", "Score"]
)

df_selected
```

### `nrows`: Used when you only want to read the first few rows of a file.

```python
# Read only the first 2 rows
df_sample = pd.read_csv(
    "../assets/Week05/Day02/Input/students.csv",
    nrows=2
)

df_sample
```
