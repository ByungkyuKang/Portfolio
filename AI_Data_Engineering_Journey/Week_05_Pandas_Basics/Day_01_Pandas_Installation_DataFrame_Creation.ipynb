## 01. DataFrame
- 2D data structure

DataFrame<br>
&nbsp;&nbsp;&nbsp;&nbsp;= table<br>
&nbsp;&nbsp;&nbsp;&nbsp;= rows + columns<br>
&nbsp;&nbsp;&nbsp;&nbsp;= Excel sheet-like structure

e.g.)
|  | Name | Age | Role| 
|---|---|---|---|
|0|Alice|25|Data Analyst|
|1|Bob|30|Data Engineer|
|2|Charlie|35|ML Engineer|

&nbsp;&nbsp;&nbsp;&nbsp;- Rows    → 0, 1, 2<br>
&nbsp;&nbsp;&nbsp;&nbsp;- Columns → Name, Age, Role

### Creating a DataFrame with a dictionary
- dictionary key   → column name
- dictionary value → column values

<i><b>dict_var</b></i> = {<br>
&nbsp;&nbsp;&nbsp;&nbsp;<i>"Column_1": [ "row_1_1", "row_1_2", "row_1_3", ...],<br>
&nbsp;&nbsp;&nbsp;&nbsp;"Column_2": [ "row_2_1", "row_2_2", "row_2_3", ...],<br>
&nbsp;&nbsp;&nbsp;&nbsp;"Column_3": [ "row_3_1", "row_3_2", "row_3_3", ...],<br></i>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;:<br>
}

pd.DataFrame( <i><b>dict_var</b></i> )

-->

|  | Column_1 | Column_2 | Column_3 | ... |
|---|---|---|---|---|
|0| row_1_1 | row_2_1 | row_3_1 | ... |
|1| row_1_2 | row_2_2 | row_3_2 | ... |
|2| row_1_3 | row_2_3 | row_3_3 | ... |


```python
import pandas as pd

# Create a dictionary with column names as keys
data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "Role": ["Data Analyst", "Data Engineer", "ML Engineer"]
}

# Create a DataFrame from the dictionary
df = pd.DataFrame(data)

df
```

### Creating a DataFrame with a list of dictionaries
- This approach is commonly used when converting JSON data or API responses into a DataFrame.

<i><b>dict_var</b></i> = [<br>
&nbsp;&nbsp;&nbsp;&nbsp;{ <i>"Column_1":"row_1_1", "Column_2":"row_2_1", "Column_3":"row_3_1"</i> },<br>
&nbsp;&nbsp;&nbsp;&nbsp;{ <i>"Column_1":"row_1_2", "Column_2":"row_2_2", "Column_3":"row_3_2"</i> },<br>
&nbsp;&nbsp;&nbsp;&nbsp;{ <i>"Column_1":"row_1_3", "Column_2":"row_2_3", "Column_3":"row_3_3"</i> },<br></i>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;:<br>
&nbsp;&nbsp;&nbsp;&nbsp;}
]

pd.DataFrame( <i><b>dict_var</b></i> )

-->

|  | Colum_1 | Colum_2 | Colum_3 | ... |
|---|---|---|---|---|
|0| row_1_1 | row_2_1 | row_3_1 | ... |
|1| row_1_2 | row_2_2 | row_3_2 | ... |
|2| row_1_3 | row_2_3 | row_3_3 | ... |


```python
import pandas as pd

# Create a list of dictionaries
students = [
    {"Name": "Alice", "Age": 25, "Score": 88},
    {"Name": "Bob", "Age": 30, "Score": 92},
    {"Name": "Charlie", "Age": 35, "Score": 79}
]

# Create a DataFrame from the list of dictionaries
df_students = pd.DataFrame(students)

df_students
```

## 02. Custom index
- Default index: 0, 1, 2, 3, ...
- but custom indexes can be set

<i>var</i> = pd.DataFrame( data, <b>index</b>=[ <i>"index_1", "index_2", ... </i> ] )

```python
import pandas as pd

# Create sample data
data = {
    "Skill": ["Python", "SQL", "Pandas"],
    "Level": ["Beginner", "Intermediate", "Beginner"]
}

# Create a DataFrame with custom index labels
df_skills = pd.DataFrame(data, index=["A", "B", "C"])

df_skills
```
