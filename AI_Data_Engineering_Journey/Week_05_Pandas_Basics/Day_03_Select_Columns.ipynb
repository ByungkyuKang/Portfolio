## Goal
- Selecting single or multiple columns within a pandas DataFrame for data subsetting.

### Today's Key Concepts
- df["column_name"]: Selecting a single column (returns a Series)
- df[["column1", "column2"]]: Selecting multiple columns (returns a DataFrame)
- Series: Understanding the 1D data structure in pandas
- DataFrame: Understanding the 2D data structure in pandas
- Checking Columns: Inspecting column names using df.columns
- Saving to Variables: Assigning a selected column to a new variable
- Basic Calculations: Performing arithmetic operations on selected columns

```python
import pandas as pd

# Read the students CSV file
df_students = pd.read_csv("../assets/Week05/Day03/Input/students.csv")

# Display the DataFrame
df_students
```

---
## 01. Checking Column names

```python
# Check column names
df_students.columns
```

```python
# Convert column names to a list
list(df_students.columns)
```

## 02. Selecting a Single Column
- To select a single column, pass the column name as a string inside square brackets.<br>
&nbsp;&nbsp;&nbsp;&nbsp;df<b>[</b> <i>"column"</i> <b>]</b>

```python
# Select one column
df_students["Name"]
```

- Selecting a single column returns a Series rather than a DataFrame.

```python
# Check the type of one selected column
name_column = df_students["Name"]

print(type(name_column))
```

---
## 02. Series
- In pandas, a Series represents a 1D (one-dimensional) data structure. Think of it as a single column pulled out from a DataFrame table.<br>

---
## 03. Selecting a Single Column as a DataFrame
- Use double square brackets.<br>
&nbsp;&nbsp;&nbsp;&nbsp;df <b>[[</b> "column"** <b>]]</b>

```python
# Check the type
name_dataframe = df_students[["Name"]]

print(type(name_dataframe))
```

- summary<br>
&nbsp;&nbsp;&nbsp;&nbsp;df_students["Name"]&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;# Series<br>
&nbsp;&nbsp;&nbsp;&nbsp;df_students[["Name"]]&nbsp;&nbsp;&nbsp;&nbsp;# DataFrame

---
## 04. Selecting multiple columns
- Use a list of columns<br>
&nbsp;&nbsp;&nbsp;&nbsp;df[ **[column1, column2, ...]** ]
&nbsp;&nbsp;-> The reason for double brackets is that the inner brackets represent a list of column names

```python
# Select multiple columns
df_students[["Name", "Score"]]
```

---
## 05. Selecting a numeric column allows you to perform calculations.
- .mean()
- .max()
- .min()

```python
# Select the Score column
scores = df_students["Score"]

# Calculate basic statistics from the Score column
print("Max score:", df_students["Score"].max())
print("Min score:", df_students["Score"].min())
print("Average score:", df_students["Score"].mean())
```

### Practice 8 — Load Model Results and Select Columns
#### Goal: Practice selecting columns from another CSV file.
##### <u>Requirements</u>
&nbsp;&nbsp;&nbsp;&nbsp;Read the following CSV file:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;../assets/Week05/Day03/Input/model_results.csv<br><br>
&nbsp;&nbsp;&nbsp;&nbsp;Select the following columns:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Model<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Accuracy<br><br>
&nbsp;&nbsp;&nbsp;&nbsp;Store the result in df_model_accuracy.<br>
&nbsp;&nbsp;&nbsp;&nbsp;Display the result.

```python
import pandas as pd
import os

dir_path = "../assets/Week05/Day03/Input/"
file_name = "model_results.csv"
full_file_path = f"{dir_path}{file_name}"

if os.path.exists(dir_path):
    os.makedirs(dir_path, exist_ok=True)

if os.path.exists(full_file_path):
    df_models = pd.read_csv(full_file_path)
    print("Successfully loaded.")
else:
    print("File not exists.")

df_model_accuracy = df_models[["Model", "Accuracy"]]
print(df_model_accuracy)
```

### Practice 9 — Calculate Average Model Accuracy
