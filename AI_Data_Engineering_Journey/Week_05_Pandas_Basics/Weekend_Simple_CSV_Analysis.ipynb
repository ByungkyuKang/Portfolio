```python
import os
import pandas as pd

# Create input and output folders
input_dir = "../assets/Week05/Weekend/Input/"
input_file = "students.csv"
input_full_path = f"{input_dir}{input_file}"

output_dir = "../assets/Week05/Weekend/Output/"
summary_file = f"{output_dir}/analysis_summary.txt"
high_score_file = f"{output_dir}/high_score_students.csv"
passed_students_file = f"{output_dir}/passed_students.csv"

os.makedirs(input_dir, exist_ok=True)
os.makedirs(output_dir, exist_ok=True)
```

```python
# Create sample student data
student_data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Age": [25, 30, 35, 28, 22, 40],
    "Score": [88, 92, 79, 85, 67, 95],
    "Department": ["Data", "Engineering", "Data", "Cloud", "Engineering", "Security"],
    "Passed": [True, True, False, True, False, True]
}

# Create a DataFrame
df_students = pd.DataFrame(student_data)

# Save the DataFrame as a CSV file
df_students.to_csv(input_full_path, index=False)

print("students.csv has been created.")
```

```python
# Read the CSV file
df = pd.read_csv(input_full_path)

# Display the DataFrame
df
```

```python
# Inspect the DataFrame
print("First 5 rows:")
print(df.head())

# Shape
print("\nShape:", df.shape)

# Columns
print("\nColumns:", list(df.columns))

# Data type
print("\nData type:")
print(df.dtypes)
```

```python
# Select important columns
df_name_score = df[ ["Name", "Score"] ]
df_name_score
```

```python
# Filter high score students
high_score_students = df[ df["Score"]>=85 ]
high_score_students
```

```python
# Filter passed students
passed_students = df[ df["Passed"] ]
passed_students
```

```python
# Calculate simple statistics
# total, average, maximum, minimum, sum
total_students = len(df)
average_score = df["Score"].mean()
max_score = df["Score"].max()
min_score = df["Score"].min()
passed_count =  df["Passed"].sum()
passed_rate = df["Passed"].sum()/total_students*100

print(f"Total students: {total_students}")
print(f"Average score: {average_score:.2f}")
print(f"Max score: {max_score}")
print(f"Min score: {min_score}")
print(f"Passed count: {passed_count}")
print(f"Passed rate: {passed_rate:.2f}%")
```

```python
# Save analysis summary to a text file
with open(summary_file, "w", encoding="utf-8") as file:
    file.write("Simple CSV Analysis Summary\n")
    file.write("===========================\n")
    file.write(f"Total students: {total_students}\n")
    file.write(f"Average score: {average_score:.2f}\n")
    file.write(f"Max score: {max_score}\n")
    file.write(f"Min score: {min_score}\n")
    file.write(f"Passed count: {passed_count}\n")
    file.write(f"Passed rate: {passed_rate:.2f}%\n")

print("Analysis summary has been saved.")
```

```python
# Read the saved summary file to confirm the result
with open(summary_file, "r", encoding="utf-8") as file:
    saved_summary = file.read()

print(saved_summary)
```

```python
# Save filtered DataFrames as CSV files
high_score_students.to_csv(high_score_file, index=False)
passed_students.to_csv(passed_students_file, index=False)

print("Filtered CSV files have been saved.")
```
